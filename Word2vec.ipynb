```python
from nltk.tokenize import word_tokenize
from gensim.models import Word2Vec
import torch
import pandas as pd

# load data
fake_data = pd.read_csv('News _dataset/Fake.csv')
true_data = pd.read_csv('News _dataset/True.csv')

# combine data
data = pd.concat([fake_data, true_data], ignore_index=True)

# preprocess data
texts = (data['title'] + " " + data['text']).tolist()
tokenized_texts = [word_tokenize(text.lower()) for text in texts]

# train word2vec model
word2vec_model = Word2Vec(sentences=tokenized_texts, vector_size=100, window=5, min_count=1, workers=4)

# get vocab
vocab = word2vec_model.wv.key_to_index
embedding_matrix = torch.zeros((len(vocab), 100))

for word, idx in vocab.items():
    embedding_matrix[idx] = torch.tensor(word2vec_model.wv[word])

# save embeddings
torch.save(embedding_matrix, 'word2vec_embeddings.pth')
```
